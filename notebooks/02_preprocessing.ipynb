```python
import pandas as pd

# Load raw dataset
df = pd.read_csv('../data/raw/air_quality_raw.csv')

# Keep only the columns we need
cols = ['Date', 'City', 'PM2.5', 'PM10', 'NO2', 'AQI', 'AQI_Bucket']
df = df[cols].copy()

print("Initial shape:", df.shape)
print("\nMissing values before cleaning:")
print(df.isnull().sum())

# Convert Date
df['Date'] = pd.to_datetime(df['Date'])

# Drop rows where PM2.5 is missing
df = df.dropna(subset=['PM2.5'])

# Fill PM10 and NO2 with median
df['PM10'] = df['PM10'].fillna(df['PM10'].median())
df['NO2'] = df['NO2'].fillna(df['NO2'].median())

# Optional: fill AQI_Bucket if missing
df['AQI_Bucket'] = df['AQI_Bucket'].fillna('Unknown')

print("\nShape after missing-value handling:", df.shape)
print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nPreview:")
print(df.head())
```

```python
def clip_outliers_iqr(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return series.clip(lower, upper)

for col in ['PM2.5', 'PM10', 'NO2']:
    df[col] = clip_outliers_iqr(df[col])

print(df[['PM2.5', 'PM10', 'NO2']].describe())
```

```python
df.to_csv('../data/clean/air_quality_clean.csv', index=False)
print("✅ Saved cleaned dataset to data/clean/air_quality_clean.csv")
print("Final shape:", df.shape)
```
