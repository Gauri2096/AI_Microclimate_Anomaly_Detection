```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
import joblib

# Load clean data
df = pd.read_csv('..data/clean/air_quality_clean.csv')

# Select features to normalize
features = ['PM2.5', 'PM10', 'NO2']

# Initialize and fit scaler
scaler = MinMaxScaler(feature_range=(0, 1))
df[features] = scaler.fit_transform(df[features])

# Save scaler for later use in calculate_index()
joblib.dump(scaler, 'models/scaler.pkl')
print("✅ Saved scaler to models/scaler.pkl")

# Verify normalization
print("\nNormalized stats (should all be 0–1):")
print(df[features].describe())

# Save normalized dataset
df.to_csv('data/normalized/air_quality_normalized.csv', index=False)
print("\n✅ Saved normalized data to data/normalized/air_quality_normalized.csv")
print("Final shape:", df.shape)
```
