```python
df = pd.read_csv('../data/raw/air_quality_raw.csv')

print("Dataset loaded successfully!")
print(f"Shape: {df.shape}")
print("\nKey columns:", df.columns.tolist())
print("\nPM2.5 stats:")
print(df['PM2.5'].describe() if 'PM2.5' in df.columns else "No PM2.5 column")
print("\nMissing values per column:")
print(df.isnull().sum())
print("\nFirst 3 rows:")
print(df.head(3))
summary = {
    'rows': len(df),
    'pm25_range': f"{df['PM2.5'].min():.1f} - {df['PM2.5'].max():.1f}",
    'has_temp': 'Temp' in df.columns,
    'has_humidity': 'RH' in df.columns or 'Humidity' in df.columns
}
print("\n✅ SUMMARY:", summary)
```
